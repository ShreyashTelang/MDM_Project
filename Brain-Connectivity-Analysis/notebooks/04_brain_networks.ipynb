# 04. Graph Theory & Brain Network Analysis
Constructs weighted NetworkX graphs, computes degree centrality, clustering coefficients, and global efficiency.

```python
import sys
sys.path.append('..')

import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
from src.graph_builder import BrainGraphBuilder
```

```python
builder = BrainGraphBuilder(threshold_value=85.0)
dummy_mat = np.random.normal(0, 0.4, (116, 116))
dummy_mat = (dummy_mat + dummy_mat.T) / 2.0

G = builder.build_networkx_graph(dummy_mat)
metrics = builder.compute_graph_metrics(G)
print(f'Active Edges: {metrics["num_edges"]}')
print(f'Network Density: {metrics["network_density"]:.4f}')
print(f'Average Clustering Coeff: {metrics["avg_clustering_coefficient"]:.4f}')
print(f'Global Efficiency: {metrics["global_efficiency"]:.4f}')
print('Top Hubs:')
for h in metrics['top_hubs']:
    print(f'  {h["region"]} ({h["network"]}): Degree = {h["degree"]:.3f}')
```
