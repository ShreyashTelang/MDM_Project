# 03. Functional Connectivity Analysis
Computes Pearson correlation coefficients and variance-stabilized Fisher r-to-z transformed connectomes.

```python
import sys
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.connectivity import FunctionalConnectivity
from config import AAL_REGIONS
```

```python
fc = FunctionalConnectivity(apply_fisher_z=True)
ts = np.random.normal(0, 1, (180, 116))
fc_mat = fc.compute_matrix(ts)

plt.figure(figsize=(8, 7))
sns.heatmap(fc_mat, cmap='coolwarm', center=0, vmin=-0.8, vmax=0.8)
plt.title('116x116 Functional Connectivity Matrix (AAL Atlas)')
plt.xlabel('AAL Region Index')
plt.ylabel('AAL Region Index')
plt.show()
```
