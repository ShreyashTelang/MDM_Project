# 02. fMRI Preprocessing & Signal Conditioning
Applies zero-phase forward-backward Butterworth bandpass filtering (0.01-0.10 Hz), linear detrending, and temporal normalization.

```python
import sys
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
from src.preprocessing import FMRIPreprocessor
```

```python
preprocessor = FMRIPreprocessor()
raw_signal = np.random.normal(0, 1, (180, 116)) + np.linspace(0, 3, 180)[:, None]
clean_signal = preprocessor.process(raw_signal)

plt.figure(figsize=(12, 4))
plt.plot(raw_signal[:, 0], label='Raw BOLD (Drift + Noise)', color='gray', alpha=0.7)
plt.plot(clean_signal[:, 0], label='Clean BOLD (0.01-0.10 Hz)', color='teal', linewidth=2)
plt.title('fMRI Signal Conditioning (Zero-Phase Butterworth Bandpass)')
plt.xlabel('Time Points (TR=2.0s)')
plt.ylabel('Amplitude')
plt.legend()
plt.show()
```
