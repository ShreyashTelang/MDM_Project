# 01. ABIDE I Dataset Exploration
This notebook loads the ABIDE I resting-state fMRI cohort and inspects demographic distributions, site balance, and AAL-116 parcellation.

```python
import sys
from pathlib import Path
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import ABIDELoader
from config import AAL_REGIONS
```

```python
loader = ABIDELoader()
pheno_df, ts_dict = loader.fetch_dataset(n_subjects=100)
summary = loader.get_dataset_summary()
print('Loaded Cohort Summary:')
for k, v in summary.items():
    print(f'  {k}: {v}')
```

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=pheno_df, x='SITE_ID', hue='DX_GROUP_MAPPED', palette='coolwarm')
plt.title('Participant Distribution by Scanning Site & Diagnosis')
plt.xlabel('Scanning Site')
plt.ylabel('Participant Count')
plt.show()
```
